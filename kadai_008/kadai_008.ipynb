# 条件分岐のif文で処理を切り替えよう

課題のコードと実行結果を記録しています。

```python
var = 15
if var % 3 == 0 and var % 5 == 0: print("FizzBuzz")
elif var % 3 == 0: print("Fizz")
elif var % 5 == 0: print("Buzz")
else: print(var)
```

```python
def fizzbuzz_output(var):
    if var % 3 == 0 and var % 5 == 0:
        return "FizzBuzz"
    elif var % 3 == 0:
        return "Fizz"
    elif var % 5 == 0:
        return "Buzz"
    else:
        return var

for var, expected in [(15, "FizzBuzz"), (9, "Fizz"), (10, "Buzz"), (7, 7)]:
    actual = fizzbuzz_output(var)
    assert actual == expected
    print(f"{var} -> {actual}")
```

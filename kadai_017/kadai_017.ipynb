```python
ADULT_AGE = 20  # グローバル定数

class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= ADULT_AGE:
            print(f"{self.name}さんは大人です。")
        else:
            print(f"{self.name}さんは大人ではありません。")

def check_humans(people):
    # people/person はこの関数内のローカル変数
    for index in range(len(people)):
        person = people[index]
        person.check_adult()
    return len(people)

human_list = [
    Human("侍太郎", 36),
    Human("侍花子", 18),
    Human("侍次郎", 20),
]

human_by_name = {person.name: person for person in human_list}
print("辞書の登録名:", list(human_by_name.keys()))
print("判定人数:", check_humans(human_list))
```
